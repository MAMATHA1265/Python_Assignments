```python
#7. SETS
#Create a set containing 5 numbers.
set={29,45,34,84,90}
print(set)
#Add a new element using add().
set.add(48)
print(set)
#Remove an element using remove().
set.remove(29)
print(set)
#Check whether an element exists in a set.
print(84 in set)
#Create two sets and find their union.
set1={29,45,34,84,90}
set2={34,84,56,79}
print(set1.union(set2))
#Create two sets and find their intersection.
set1={29,45,34,84,90}
set2={34,84,56,79}
print(set1.intersection(set2))
#Create two sets and find their difference.
set1={29,45,34,84,90}
set2={34,84,56,79}
print(set1.difference(set2))
#Create a set containing duplicate numbers and observe the result.
s={23,22,15,63,22,15,98,59,63}
print(s)

#Given two sets of student names, find students present in both sets.
set1={"neha","adya","mahan","krishiv","megha","krishan"}
set2={"pooja","krishiv","banny","raju","krishan"}
print(set1.intersection(set2))
#Given two sets of students, find students who are present in the first class but not in the second class.

set1={"neha","adya","mahan","krishiv","megha","krishan"}
set2={"pooja","krishiv","banny","raju","krishan"}
print(set1.difference(set2))
```
