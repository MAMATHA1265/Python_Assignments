```python
#5. DICTIONARIES
#Use the following dictionary for Questions 1–8:
student = {
    "name": "Rahul",
    "age": 20,
    "marks": 75,
    "city": "Delhi"
}
#Print the student's name.
print(student["name"])
#Print the student's marks.
print(student["marks"])
#Add a new key called "course".
student["course"]="GenAI"
print(student)
#Update the student's marks to 85.
student["city"]="Delhi"
print(student)
#Add an "email" key.
student["email"]="Rahul@gmail.com"
print(student)
#Remove the "age" key.
student.pop("age")
print(student)
#Print all keys using keys().
for i in student.keys():
   print(i)
#Print all values using values().
for i in student.values():
   print(i)
#Additional Dictionary Practice
#Create a dictionary containing 5 students and their marks. Find the student with the highest marks.
student_marks={"neha":56,"krish":69,"vedha":53,"megha":48,"jathin":92}
highest_marks=max(student_marks,key=student_marks.get)
print("highest_marks",highest_marks,"with a score of:",student_marks[highest_marks])
        
#Create a dictionary containing 5 products and their prices. Find the most expensive product.
product_price={"laptop":25000,"tab":5000,"mobile":10000,"earphones":2000,"smartwatch":9000}
expensive_product=max(product_price,key=product_price.get)
print("expensive_product:",expensive_product)
#Create a dictionary containing student names and marks. Count how many students passed.
student_marks={"neha":56,"krish":34,"vedha":37,"megha":48,"jathin":92}
count=0
for marks in student_marks.values():
  if marks>=40:
     count=count+1
print("passed_students:",count)
#Create a dictionary containing student names and marks. Count how many students failed.
student_marks={"neha":56,"krish":34,"vedha":37,"megha":48,"jathin":92}
count=0
for marks in student_marks.values():
  if marks<=40:
     count=count+1
print("failed_students:",count)
#Create a dictionary containing employee names and salaries. Find the highest salary.
emp_sal={"navya":21000,"pooja":20000,"ekta":23000,"mruna":30000,"mahan":38000}
highest_sal=max(emp_sal,key=emp_sal.get)
print("highest_sal:",highest_sal)
#Create a dictionary of products and prices. Calculate the total value of all products.
product_price={"laptop":25000,"tab":5000,"mobile":10000,"earphones":2000,"smartwatch":9000}
total_value=sum(product_price.values())
print("total_value",total_value)
#Create a dictionary containing city names and temperatures. Find the city with the highest temperature.
city_temp={"chennai":25,"hyderabad":34,"delhi":37,"banglore":40,"pune":42}
highest_temp=max(city_temp,key=city_temp.get)
print("highest_temp:",highest_temp,"with a temp of:",city_temp[highest_temp])
```
